## Practical No. 6: Adagrad, RMSprop and Adadelta

**Aim:** Implement the Adagrad method with application, RMSprop and Adadelta.

```python
import math

x = 10.0
learning_rate = 1.0
epsilon = 1e-8
G = 0.0

print("AdaGrad Optimization")
print("--------------------")

for i in range(10):
    gradient = 2 * x

    G = G + gradient ** 2

    x = x - (learning_rate / (math.sqrt(G) + epsilon)) * gradient

    print("Iteration", i + 1,
          ": x =", round(x, 4),
          "f(x) =", round(x ** 2, 4))

print("\nFinal x =", round(x, 4))
print("Minimum value =", round(x ** 2, 4))

x = 10.0
learning_rate = 0.1
decay = 0.9
E = 0.0

print("\nRMSprop Optimization")
print("--------------------")

for i in range(10):
    gradient = 2 * x

    E = decay * E + (1 - decay) * gradient ** 2

    x = x - (learning_rate / (math.sqrt(E) + epsilon)) * gradient

    print("Iteration", i + 1,
          ": x =", round(x, 4),
          "f(x) =", round(x ** 2, 4))

x = 10.0
rho = 0.95
eps = 1e-6

Eg = 0.0
Edx = 0.0

print("\nAdadelta Optimization")
print("---------------------")

for i in range(10):
    gradient = 2 * x

    Eg = rho * Eg + (1 - rho) * gradient ** 2

    dx = -(math.sqrt(Edx + eps) / math.sqrt(Eg + eps)) * gradient

    Edx = rho * Edx + (1 - rho) * dx ** 2

    x = x + dx

    print("Iteration", i + 1,
          ": x =", round(x, 4),
          "f(x) =", round(x ** 2, 4))
```
